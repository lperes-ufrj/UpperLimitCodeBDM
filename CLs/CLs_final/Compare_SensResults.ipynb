```python
import numpy as np
from array import *
import matplotlib.pyplot as plt
import os
import time
import glob
import dunestyle.matplotlib as dunestyle
import pandas as pd
```

```python
# Specify the folder where the files are located
folder_path_1p4 = 'plots/20260512_101630/'
folder_path_0p4 = 'plots/20260619_061009/'

# Use glob to find all files matching the pattern in the specified folder
pattern_1 = os.path.join(folder_path_1p4, 'limits*.dat')
pattern_2 = os.path.join(folder_path_0p4, 'limits*.dat')

# Get all files matching the pattern
files_1p4 = sorted(glob.glob(pattern_1))
files_0p4 = sorted(glob.glob(pattern_2))

print(f"Found {len(files_1p4)} files in folder 1: {files_1p4}")
print(f"Found {len(files_0p4)} files in folder 2: {files_0p4}")
# Move the first 4 files to the last place
if len(files_1p4) > 4:
    files_1p4 = files_1p4[4:] + files_1p4[:4]
if len(files_0p4) > 4:
    files_0p4 = files_0p4[4:] + files_0p4[:4]

# Initialize a list of 5 empty lists, one for each line
limits_1p4 = [[] for _ in range(5)]
limits_0p4 = [[] for _ in range(5)]

# Read each file
for file in files_1p4 + files_0p4:
    with open(file, 'r') as f:
        # Read each line and store it in the appropriate list
        for i, line in enumerate(f.readlines()):
            number = float(line.strip())  # Convert the line to an integer
            if file in files_1p4:
                limits_1p4[i].append(number)
            else:
                limits_0p4[i].append(number)
```

```python
# Print each list of numbers
for i, values in enumerate(limits_1p4, start=1):
    print(f"Numbers from line {i} across all files: {values}")

# Print each list of numbers
for i, values in enumerate(limits_0p4, start=1):
    print(f"Numbers from line {i} across all files: {values}")
```

```python
# Table comparing the two sensitivity folders
sigma_labels = ["+2sigma", "+1sigma", "central", "-1sigma", "-2sigma"]
sample_labels_1p4 = [
    os.path.splitext(os.path.basename(file))[0].replace("limits_", "")
    for file in files_1p4
]
sample_labels_0p4 = [
    os.path.splitext(os.path.basename(file))[0].replace("limits_", "")
    for file in files_0p4
]

if sample_labels_1p4 != sample_labels_0p4:
    raise ValueError(
        "The two sensitivity folders are not ordered the same way:\n"
        f"folder 1: {sample_labels_1p4}\n"
        f"folder 2: {sample_labels_0p4}"
    )

sensitivity_1p4_table = pd.DataFrame(
    limits_1p4,
    index=sigma_labels,
    columns=sample_labels_1p4,
)
sensitivity_0p4_table = pd.DataFrame(
    limits_0p4,
    index=sigma_labels,
    columns=sample_labels_0p4,
)
ratio_table = sensitivity_0p4_table / sensitivity_1p4_table

print(f"Sensitivity 1: {folder_path_1p4}")
print(sensitivity_1p4_table.to_string(float_format=lambda value: f"{value:.6e}"))

print(f"\nSensitivity 2: {folder_path_0p4}")
print(sensitivity_0p4_table.to_string(float_format=lambda value: f"{value:.6e}"))

print("\nRatio: sensitivity 2 / sensitivity 1")
print(ratio_table.to_string(float_format=lambda value: f"{value:.6f}"))
```
